```python
class MyAnimals:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __sub__(self, other):
        return f"Конфлікт між {self.name} та {other.name}"

    def __gt__(self, other):
        return self.age > other.age

    def __len__(self):
        return len(self.name)

a = MyAnimals("Рекс", 5)
b = MyAnimals("Мурзик", 3)

print(a - b)
print(f"Чи старший a за b: {a > b}")
print(f"Довжина імені: {len(a)}")
```

```python
class MyAnimals:
    @classmethod
    def is_valid_name(cls, name):
        return name[0].isupper()

print(f"Результат перевірки 'Рекс': {MyAnimals.is_valid_name('Рекс')}")
print(f"Результат перевірки 'сірко': {MyAnimals.is_valid_name('сірко')}")
```

```python
class MyAnimals:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @property
    def age_in_months(self):
        return self.age * 12

a1 = MyAnimals("Мурчик", 5)
print(f"Вік у місяцях: {a1.age_in_months}")
```

```python
class MyAnimals:
    def __init__(self, animals, name, age):
        self.animals = animals
        self.name = name
        self.age = age

    @staticmethod
    def call_pet():
        return "Тваринка біжить до нас!"

a1 = MyAnimals("Кіт", "Мурчик", 5)
print(f"Виклик з об'єкта: {a1.call_pet()}")
print(f"Виклик з класу: {MyAnimals.call_pet()}")
```

```python
class MyAnimals:
    total_animals = 0
    total_animal_types = set()

    def __init__(self, animals, name, age):
        self.animals = animals
        self.name = name
        self.age = age
        MyAnimals.total_animals += 1
        MyAnimals.total_animal_types.add(animals)

for i in range(3):
    MyAnimals("Папуга", f"Кеша_{i}", i+1)

print(f"Загальна кількість: {MyAnimals.total_animals}")
```

```python
class MyAnimals:
    def __init__(self, animals, name, age):
        self.animals = animals
        self.name = name
        self.age = age
    
    def speak(self):
        if self.animals == "Кіт": return "Мяууу"
        if self.animals == "Собака": return "Гавв"
        return "Не розмовляє"
    
    def add_one_year(self):
        self.age += 1
        return f"{self.name} подорослішав(ла). Тепер вік: {self.age}"

a1 = MyAnimals("Рибки", "Золота_Рибка", 1)
a2 = MyAnimals("Кіт", "Мурчик", 5)
a3 = MyAnimals("Собака", "Бровко", 2)

for obj in [a1, a2, a3]:
    print(obj.speak())
    print(obj.add_one_year())
```

```python
class MyAnimals:
    def __init__(self, animals, name, age, mass=None) -> None:
        self.animals = animals
        self.name = name
        self.age = age
        self.mass = 0 if mass is None else mass

a1 = MyAnimals("Рибки", "Неони", 1)
a2 = MyAnimals("Кіт", "Мурка", 5, 10)
a3 = MyAnimals(animals="Собака", name="Рекс", age=2)

a1.name, a2.name, a3.name = "Золота_Рибка", "Мурчик", "Бровко"

for obj in [a1, a2, a3]:
    print(f"Тварина: {obj.animals}, Ім'я: {obj.name}")
```
